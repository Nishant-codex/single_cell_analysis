### Inhibitory

```python
cols = cols_ephys
save_dtype = 'ephys'
ei_type = 'inh'
adj_rand_score_inh_ephys = {}
adj_MI_inh_ephys = {}
for cond in ['d1','d2','m1-ago']:
    print(cond)
    df_acsf,df_drug = return_acsf_and_drug(df_above_threshold[df_above_threshold.ei_labels==0.0],cond=[cond],joint=False)
    common_exps = set(df_acsf.exp_name) & set(df_drug.exp_name)
    df_acsf = df_acsf.drop_duplicates('exp_name')
    df_acsf = df_acsf[df_acsf.exp_name.isin(common_exps)]
    df_drug = df_drug.drop_duplicates('exp_name')
    df_drug = df_drug[df_drug.exp_name.isin(common_exps)]

    data_acsf = normalize(np.vstack(df_acsf[cols].to_numpy()),axis=0)
    data_drug = normalize(np.vstack(df_drug[cols].to_numpy()),axis=0)

    lables_acsf,mapper,reducer = plot_UMAP_clusters_single(data_acsf,
                                                                20,
                                                                0.1,
                                                                random_state =30,
                                                                res_louvain = 1.,
                                                                c_list =CUSTOM_PAL_SORT_3,
                                                                size=100,
                                                                norm = False,
                                                                save = False,
                                                                savepath =  savepath + cond+'/' +save_dtype+'/'+save_dtype+'_cluster_on_diff_'+ei_type+'.pdf',
                                                                figsize=(3,3)
                                                                )

    lables_diff,mapper,reducer = plot_UMAP_clusters_single((data_drug-data_acsf)/data_acsf,
                                                                20,
                                                                0.1,
                                                                random_state =30,
                                                                res_louvain = 1.,
                                                                c_list =CUSTOM_PAL_SORT_3,
                                                                size=100,
                                                                norm = False,
                                                                save = False,
                                                                savepath =  savepath + cond+'/' +save_dtype+'/'+save_dtype+'_cluster_on_diff_'+ei_type+'.pdf',
                                                                figsize=(3,3)
                                                                )

    df_temp_acsf,df_temp_drug  = return_acsf_and_drug(df_above_threshold,cond=['d1'],joint=False)
    exc_acsf = df_temp_acsf[df_temp_acsf.ei_labels==1] 
    exc_acsf = exc_acsf.drop_duplicates('exp_name')
    # exc_acsf.reset_index(True)
    exc_drug = df_temp_drug[df_temp_drug.ei_labels==1]
    exc_drug = exc_drug.drop_duplicates('exp_name')
    # exc_acsf.reset_index(True) 

    adj_rand_score_inh_ephys[cond] = adjusted_rand_score(lables_acsf,lables_diff)
    adj_MI_inh_ephys[cond]         = adjusted_mutual_info_score(lables_acsf,lables_diff)


    df_new_spikes = df_acsf[cols_spikes].reset_index(drop=True) - df_drug[cols_spikes].reset_index(drop=True)/df_acsf[cols_spikes].reset_index(drop=True)
    df_new_spikes['labels'] = lables_diff
    sns.pairplot(df_new_spikes,hue='labels',palette=CUSTOM_PAL_SORT_3)
    plt.savefig(savepath + cond+'/' +save_dtype+'/'+save_dtype+'_pairplot_on_diff_spikes_'+ei_type+'.pdf')
    plt.show()

    df_new_thrs = df_acsf[cols_thrs].reset_index(drop=True) - df_drug[cols_thrs].reset_index(drop=True)/df_acsf[cols_thrs].reset_index(drop=True)
    df_new_thrs['labels'] = lables_diff
    sns.pairplot(df_new_thrs,hue='labels',palette=CUSTOM_PAL_SORT_3)
    plt.savefig(savepath + cond+'/' +save_dtype+'/'+save_dtype+'_pairplot_on_diff_thrs_'+ei_type+'.pdf')
    plt.show()

    df_new_wf = df_acsf[cols_wf].reset_index(drop=True) - df_drug[cols_wf].reset_index(drop=True)/df_acsf[cols_wf].reset_index(drop=True)
    df_new_wf['labels'] = lables_diff
    sns.pairplot(df_new_wf,hue='labels',palette=CUSTOM_PAL_SORT_3)
    plt.savefig(savepath + cond+'/' +save_dtype+'/'+save_dtype+'_pairplot_on_diff_wf_'+ei_type+'.pdf')
    plt.show()


```
